## 4 입력 → 계산 → 출력

속력이 일정한 물체가 이동한 거리를 계산합니다. 이번 예제에서는 속력과 시간 모두 0 이상이고, 가속도는 다루지 않습니다.

| 코드의 이름 | 의미 | 단위 |
| --- | --- | --- |
| `speed_mps` | 일정한 속력 | m/s |
| `time_s` | 이동 시간 | s |
| `distance_m` | 이동 거리 | m |

`=`는 오른쪽의 계산 결과를 왼쪽 이름에 연결합니다. `distance_m = speed_mps * time_s`에서 값을 계산한 뒤 `print`로 확인합니다. 자료형과 할당 규칙은 02 수업에서 더 자세히 다룹니다.

```text
speed_mps ──┐
            ├── 곱하기 ── distance_m ── print
 time_s ────┘
```

````python
speed_mps = 2
time_s = 3
distance_m = speed_mps * time_s

print("speed_mps =", speed_mps, "m/s")
print("time_s =", time_s, "s")
print("distance_m =", distance_m, "m")
````

## 5 현재 값을 그림으로 확인하기

다음 보조 셀은 입력→결과 개념도와 거리-시간 그래프를 그립니다. 함수·리스트·반복문의 세부 문법은 후속 수업에서 배웁니다. 지금은 이 셀을 그대로 실행하고 실제로 바꿀 입력은 위의 `speed_mps`, `time_s`입니다.

그림의 이름은 코드 식별자이며 단위는 두 언어에서 동일하게 사용합니다.

````python
def show_distance_flow(speed, duration):
    """Provided visualization helper. The model assumes constant speed >= 0."""
    if speed < 0 or duration < 0:
        raise ValueError("Use non-negative speed and time in this lesson.")
    distance = speed * duration
    fig, axes = plt.subplots(1, 2, figsize=(11, 3.6))
    flow, graph = axes
    flow.set(xlim=(0, 10), ylim=(0, 6))
    flow.axis("off")
    box = dict(boxstyle="round,pad=0.6", facecolor="#eef2ff", edgecolor="#334155")
    flow.text(2, 4.5, f"speed_mps = {speed:g} m/s", ha="center", bbox=box)
    flow.text(2, 1.5, f"time_s = {duration:g} s", ha="center", bbox=box)
    flow.text(7, 3, f"distance_m\n{distance:g} m", ha="center", va="center", bbox=box)
    for y in (4.5, 1.5):
        flow.annotate("", xy=(5.7, 3), xytext=(4, y),
                      arrowprops=dict(arrowstyle="->", color="#334155"))
    flow.set_title("distance_m = speed_mps * time_s")
    times = [i * max(5, duration) / 20 for i in range(21)]
    distances = [speed * t for t in times]
    graph.plot(times, distances, color="#175cd3", label="d = v * t")
    graph.scatter([duration], [distance], color="#b42318", marker="s", label="(time_s, distance_m)")
    graph.set(xlabel="t (s)", ylabel="d (m)")
    graph.grid(alpha=0.25)
    graph.legend(fontsize=8)
    fig.tight_layout()
    plt.show()
    return {"speed_mps": speed, "time_s": duration, "distance_m": distance}
````

왼쪽 그림은 현재 입력 두 개와 계산 결과를 보여 줍니다. 오른쪽 선은 같은 속력에서 시간이 늘어날 때의 거리이고, 네모 점은 선택한 시간의 결과입니다. 그래프의 축과 단위를 함께 읽으세요.

````python
shown_state = show_distance_flow(speed_mps, time_s)
print(shown_state)
````

## 7 자가 점검 준비

다음 함수는 교사가 제공한 점검 도구입니다. 그대로 실행하세요. 숫자 입력과 계산 결과가 맞으면 `True`, 맞지 않으면 `False`를 반환합니다. `PASS`는 이 연습문제의 피드백이며 성적 인증이 아닙니다.

````python
def check_distance_result(speed, duration, answer):
    """Provided self-check helper; PASS is practice feedback, not a course grade."""
    import math
    numbers = (speed, duration, answer)
    if any(not isinstance(value, (int, float)) or isinstance(value, bool) for value in numbers):
        return False
    if any(not math.isfinite(value) for value in numbers):
        return False
    return speed >= 0 and duration >= 0 and math.isclose(
        answer, speed * duration, rel_tol=1e-9, abs_tol=1e-9
    )
````
